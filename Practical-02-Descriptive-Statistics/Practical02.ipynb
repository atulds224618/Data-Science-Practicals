# Q2 — Mean, Median, Minimum and Maximum of students

```python
mean_students = df["students"].mean()
median_students = df["students"].median()
min_students = df["students"].min()
max_students = df["students"].max()

print("Mean:", mean_students)
print("Median:", median_students)
print("Minimum:", min_students)
print("Maximum:", max_students)
```

# Q4 — Histogram of beauty

```python
plt.hist(df["beauty"], bins=20, edgecolor="black")

plt.xlabel("Beauty Score")
plt.ylabel("Frequency")
plt.title("Distribution of Beauty Scores")

plt.show()
```

# Q5 — Does average beauty differ by gender?

```python
gender_beauty = df.groupby("gender")["beauty"].agg(
    mean="mean",
    standard_deviation="std"
)

gender_beauty
```

# Q6 — Percentage of males and females that are tenured

```python
pd.crosstab(df["gender"], df["tenure"])
```

```python
tenure_percentage = pd.crosstab(
    df["gender"],
    df["tenure"],
    normalize="index"
) * 100

tenure_percentage
```

```python
tenure_percentage = pd.crosstab(
    df["gender"],
    df["tenure"],
    normalize="index"
) * 100

print("Percentage of tenure status by gender:")
print(tenure_percentage)
```
