A professional basketball team wants to compare its performance with that of players in a regional league. 2. The pros are known to have a historic mean of 12 points per game with a standard deviation of 5.5. 3. A group of 36 regional players recorded on average 10.7 points per game. 4. The pro coach would like to know whether his professional team scores on average are different from that of the regional players. State the null hypothesis Null Hypo: ("The mean point of the regional players is not different from the historic mean") Alternative Hypo: ("The mean point of the regional players is different from the historic mean")


## Step 1 — Import libraries + load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
df = pd.read_csv("DATA/teaching_ratings_20.csv")
df.head()
```

## Q1. Probability of receiving an evaluation score greater than 4.5

```python
total_observations = len(df)

greater_than_4_5 = (df["eval"] > 4.5).sum()

probability = greater_than_4_5 / total_observations

print("Total observations:", total_observations)
print("Evaluations greater than 4.5:", greater_than_4_5)
print("Probability:", probability)
```

```python
percentage = probability * 100

print("Percentage:", percentage, "%")
```

## Q2. Probability of receiving an evaluation score greater than 3.5 and less than 4.2

```python
between_3_5_and_4_2 = ((df["eval"] > 3.5) & (df["eval"] < 4.2)).sum()

print("Evaluations between 3.5 and 4.2:", between_3_5_and_4_2)
```

```python
probability_q2 = between_3_5_and_4_2 / len(df)

print("Probability:", probability_q2)
```

```python
percentage_q2 = probability_q2 * 100

print("Percentage:", percentage_q2, "%")
```


# State the hypotheses

## Null Hypothesis (H₀):

### The mean points of the regional players is not different from the historic mean.

    \[
       H_0:\mu=12
    \]

## Alternative Hypothesis (H₁):

### The mean points of the regional players is different from the historic mean.

     \[
       H_1:\mu\neq12
     \]
### Because we're checking for a difference in either direction, this is a two-tailed test.

```python
mu = 12
sigma = 5.5
sample_mean = 10.7
n = 36

z = (sample_mean - mu) / (sigma / np.sqrt(n))

print("Z-score:", z)
```

```python
p_value = 2 * norm.cdf(-abs(z))

print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```


## Final conclusion for your practical

### The sample mean of the regional players is 10.7 points per game, compared with the historic mean of 12 points per game. However, the two-tailed test gives a p-value of approximately 0.156, which is greater than 0.05. Therefore, there is not enough statistical evidence to conclude that the regional players' average score is different from the historic professional mean.
