# Q1. Regression with T-Test Using the teachers' rating dataset, does gender affect teaching evaluation rates?

## Step 1 — Convert gender into a regression variable

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import statsmodels.api as sm
import statsmodels.formula.api as smf
df = pd.read_csv("DATA/teaching_ratings_20.csv")

df.head()
df["female"] = (df["gender"] == "female").astype(int)
```

## Step 2 — Create X and Y

```python
X = df["female"]
y = df["eval"]
```

## Step 3 — Add constant

```python
X = sm.add_constant(X)
```

## Step 4 — Fit the regression

```python
model = sm.OLS(y, X).fit()
```

## Step 5 — Display the complete regression result

```python
print(model.summary())
```

```python
p_value = model.pvalues["female"]

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("There is a significant difference in teaching evaluation rates by gender.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no significant difference in teaching evaluation rates by gender.")
```

# Q2. Regression with ANOVA Using the teachers' rating dataset, does beauty score for instructors differ by age?

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 40, 55, 100],
    labels=["Below 40", "40-55", "Above 55"]
)
```

```python
print(df[["age", "age_group", "beauty"]].head())
```

```python
model_anova = smf.ols(
    "beauty ~ C(age_group)",
    data=df
).fit()
```

```python
anova_table = sm.stats.anova_lm(model_anova, typ=2)

print(anova_table)
```

```python
p_value = anova_table.loc["C(age_group)", "PR(>F)"]

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("Beauty score differs significantly across age groups.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no significant difference in beauty score across age groups.")
```

# Q3. Regression — Beauty vs Evaluation Using the teachers' rating dataset, is teaching evaluation score correlated with beauty score?

```python
X = df["beauty"]
y = df["eval"]
```

```python
X = sm.add_constant(X)
```

```python
model_beauty = sm.OLS(y, X).fit()
```

```python
print(model_beauty.summary())
```

```python
p_value = model_beauty.pvalues["beauty"]

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("There is a significant relationship between beauty score and teaching evaluation score.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no significant relationship between beauty score and teaching evaluation score.")
```
