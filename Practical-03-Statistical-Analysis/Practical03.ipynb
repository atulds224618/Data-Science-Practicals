# Question 1 Calculate the percentage of visible minorities who are tenured professors. Will you say that tenure status differed if a teacher was a visible minority?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_excel("DATA/Student_Rating_Dataset_120_Entries.xlsx")
df.head()
```

# Question 2 Does average age differ by tenure? Produce the means and standard deviations for both tenured and untenured professors.

```python
age_by_tenure = df.groupby("tenure")["age"].agg(["mean", "std"])

print(age_by_tenure)
```

# Question 3 Which graph works better and plot that graph for the age variable?

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=10, edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Professors' Age")

plt.show()
```

# Question 4 Define the difference between pyplot.bar and pyplot.barh. Also plot the graph for the gender variable.

```python
gender_counts = df["gender"].value_counts()

plt.figure(figsize=(6, 4))

plt.bar(gender_counts.index, gender_counts.values, edgecolor="black")

plt.xlabel("Gender")
plt.ylabel("Number of Professors")
plt.title("Number of Professors by Gender")

plt.show()
```

# Question 5 What is the median evaluation score for tenured professors?

```python
tenured_professors = df[df["tenure"] == "yes"]

median_eval = tenured_professors["eval"].median()

print("Median evaluation score for tenured professors:", median_eval)
```
