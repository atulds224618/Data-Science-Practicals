# CELL 1 — Import Libraries

```python
import pandas as pd
import numpy as np
```

# CELL 2 — Load Your Dataset

```python
df = pd.read_csv("DATA/student_performance_extended_20.csv")
```

# CELL 3 — Display the DataFrame

```python
df.head()
```

# CELL 4 — Check the Entire DataFrame Structure

```python
df.info()
```

# CELL 5 — Find DataFrame Shape

```python
print("DataFrame Shape:", df.shape)
```

# CELL 6 — Separate Numerical Data

```python
numeric_df = df[
    [
        "student_ID",
        "study_hours",
        "attendance_percent",
        "sleep_hours",
        "previous_grade",
        "final_exam_score"
    ]
]
numeric_df.head()
```

# CELL 7 — Check for Missing Values

```python
numeric_df.isnull().sum()
```

# CELL 8 — Convert DataFrame → NumPy

```python
arr = numeric_df.to_numpy()
print(arr)
```

# CELL 9 — NumPy Shape

```python
print("Shape:", arr.shape)
```

# CELL 10 — NumPy Size

```python
print("Size:", arr.size)
```

# CELL 11 — NumPy Dimension

```python
print("Dimension:", arr.ndim)
```

# CELL 12 — Create Matrix A

```python
A = arr[:2, :2]
print("Matrix A:")
print(A)
```

# CELL 13 — Create Matrix B

```python
B = arr[2:4, :2]
print("Matrix B:")
print(B)
```

# CELL 14 — Addition

```python
addition = A + B

print("Addition:")
print(addition)
```

# CELL 15 — Subtraction

```python
subtraction = A - B

print("Subtraction:")
print(subtraction)
```

# CELL 16 — Multiplication

```python
multiplication = A * B

print("Multiplication:")
print(multiplication)
```

# CELL 17 — Product / Matrix Multiplication

```python
product = A @ B

print("Matrix Product:")
print(product)
```

# CELL 18 — Transpose

```python
transpose = A.T

print("Transpose of A:")
print(transpose)
```

# CELL 19 — Determinant

```python
determinant = np.linalg.det(A)

print("Determinant of A:", determinant)
```

# CELL 20 — Inverse

```python
if determinant != 0:
    inverse = np.linalg.inv(A)
    print("Inverse of A:")
    print(inverse)
else:
    print("Inverse does not exist because determinant is zero.")
```

# CELL 21 — View Categorical Data

```python
categorical_df = df[
    [
        "gender",
        "parental_education",
        "internet_access",
        "extracurricular_activities",
        "part_time_job",
        "final_grade"
    ]
]

categorical_df.head()
```

# CELL 22 — One-Hot Encoding

```python
encoded_df = pd.get_dummies(categorical_df)
encoded_df.head()
```

# CELL 23 — Show Before and After

```python
print("Before One-Hot Encoding:")
print(categorical_df.head())

print("\nAfter One-Hot Encoding:")
print(encoded_df.head())
```
