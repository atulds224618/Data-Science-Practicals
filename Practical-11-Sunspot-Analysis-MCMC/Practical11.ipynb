# Step 1: Loading and Visualizing the Monthly Mean Sunspot Dataset

## **Objective:** To load the monthly mean total sunspot number dataset and visualize the variation in sunspot activity over time.

## The dataset contains monthly sunspot observations collected from January 1749 to November 2018.

## We use the following functions:

### - `pd.read_csv()` loads the dataset directly from its online source.

### - `sep=";"` specifies the semicolon separator used in the file.

### - `pd.to_datetime()` combines the year and month into a date format.

### - `plt.plot()` creates a time-series line graph.

### The graph helps us observe how sunspot activity increases and decreases over time and identify its repeating solar cycles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Official SILSO monthly sunspot dataset
url = "DATA/SN_m_tot_V2.0.csv"

# Load dataset
columns = [
    "year",
    "month",
    "decimal_date",
    "sunspot_number",
    "standard_deviation",
    "observations",
    "marker"
]

df = pd.read_csv(url, sep=";", header=None, names=columns)

# Keep the assignment's date range:
# January 1749 to November 2018
df = df[
    ((df["year"] > 1749) |
     ((df["year"] == 1749) & (df["month"] >= 1))) &
    ((df["year"] < 2018) |
     ((df["year"] == 2018) & (df["month"] <= 11)))
].copy()

# Create a proper date column
df["date"] = pd.to_datetime(
    dict(year=df["year"], month=df["month"], day=1)
)

# Display first five records
display(df.head())

# Plot sunspot activity over time
plt.figure(figsize=(15, 6))

plt.plot(
    df["date"],
    df["sunspot_number"],
    linewidth=0.8
)

plt.title("Monthly Mean Total Sunspot Number (1749–2018)")
plt.xlabel("Year")
plt.ylabel("Monthly Mean Sunspot Number")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Display dataset information
print("Total observations:", len(df))
print("Start date:", df["date"].min())
print("End date:", df["date"].max())
```

# Step 2: Modeling Sunspot Counts Using a Gamma Distribution

## **Objective:** To model monthly sunspot observations using a Gamma probability distribution, with a new cycle beginning every 12 years.

## The Gamma distribution is a continuous probability distribution commonly used to represent positive, skewed numerical data.

## It is described using two parameters:

### - **Shape parameter (a or α):** Controls the shape of the distribution.

### - **Scale parameter (b or θ):** Controls the spread of the distribution.

### For this practical, we divide the observations into groups of 12 years. Since each year contains 12 months, one cycle contains 144 monthly observations.

## The cycle number is calculated using:

### `cycle = (year - 1749) // 12`

## A separate Gamma distribution is fitted to the positive sunspot observations in each cycle.

## **Note:** The Gamma distribution is defined for positive values, so zero sunspot observations are excluded from fitting. The 12-year grouping is the assignment's modeling assumption; it is not the exact physical duration of every solar cycle.

```python
from scipy.stats import gamma

# Create a copy of the dataset
gamma_df = df.copy()

# Assign a new cycle every 12 years
gamma_df["cycle"] = (
    (gamma_df["year"] - 1749) // 12
) + 1

# Store Gamma distribution parameters
cycle_results = []

# Fit a Gamma distribution separately for each cycle
for cycle_number, group in gamma_df.groupby("cycle"):

    # Remove zero values because Gamma distribution requires positive data
    positive_values = group.loc[
        group["sunspot_number"] > 0,
        "sunspot_number"
    ]

    # Fit Gamma distribution
    shape, loc, scale = gamma.fit(
        positive_values,
        floc=0
    )

    cycle_results.append({
        "Cycle": cycle_number,
        "Observations": len(group),
        "Shape (a)": shape,
        "Scale (b)": scale
    })

# Display fitted parameters
gamma_results = pd.DataFrame(cycle_results)

display(gamma_results.round(3))

# Plot the Gamma distribution for the first cycle
first_cycle = gamma_df[gamma_df["cycle"] == 1]

positive_values = first_cycle.loc[
    first_cycle["sunspot_number"] > 0,
    "sunspot_number"
]

shape, loc, scale = gamma.fit(positive_values, floc=0)

# Generate values for the Gamma probability density curve
x = np.linspace(0, positive_values.max() * 1.2, 300)
pdf = gamma.pdf(x, a=shape, loc=0, scale=scale)

plt.figure(figsize=(9, 5))

plt.hist(
    positive_values,
    bins=20,
    density=True,
    alpha=0.6,
    label="Observed Sunspot Counts"
)

plt.plot(
    x,
    pdf,
    linewidth=2,
    label="Fitted Gamma Distribution"
)

plt.title("Gamma Distribution Fit - First 12-Year Cycle")
plt.xlabel("Sunspot Number")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("First cycle shape parameter (a):", round(shape, 3))
print("First cycle scale parameter (b):", round(scale, 3))
```

# Conclusion

## The monthly sunspot observations were divided into groups of 12 years, and a Gamma distribution was fitted separately to the positive observations in each group. The shape and scale parameters were calculated for every cycle. The fitted Gamma probability density curve was also plotted for the first cycle to compare the theoretical distribution with the observed sunspot counts. This step provides the probability model required for further statistical and MCMC analysis.

# Step 4: Applying Markov Chain Monte Carlo (MCMC)

## **Objective:** To estimate the shape parameter (a) and scale parameter (b) of the Gamma distribution using MCMC sampling.

## MCMC is a statistical method used to generate samples from a probability distribution, particularly when direct sampling is difficult.

## In this practical, we use the Metropolis-Hastings algorithm.

### - Initial shape parameter: `a = 4`

### - Initial scale parameter: `b = 10`

### - The Gamma distribution models the positive sunspot observations.

### - A proposal mechanism generates new candidate values of a and b.

### - The candidate values are accepted or rejected according to their posterior probability.

## We generate 5,000 MCMC iterations and examine three portions of the resulting chain:

### 1. First 50 samples: Shows the initial behavior of the chain.

### 2. All samples: Shows the complete sampling process.

### 3. Last 50 samples: Shows the behavior near the end of the chain.

## The trace plots help us observe how the estimated parameters change throughout the MCMC process.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gammaln

# Use positive sunspot observations
data = df.loc[
    df["sunspot_number"] > 0,
    "sunspot_number"
].to_numpy()

# Initial values
initial_a = 4
initial_b = 10

# Number of MCMC iterations
n_iterations = 5000

# Random number generator
rng = np.random.default_rng(42)

# Log posterior function
def log_posterior(log_params):

    # Convert log parameters into positive values
    a = np.exp(log_params[0])
    b = np.exp(log_params[1])

    # Gamma log-likelihood
    n = len(data)

    log_likelihood = (
        (a - 1) * np.sum(np.log(data))
        - np.sum(data) / b
        - n * gammaln(a)
        - n * a * np.log(b)
    )

    # Broad log-normal priors
    log_prior = (
        -0.5 * ((log_params[0] - np.log(4)) / 2) ** 2
        -0.5 * ((log_params[1] - np.log(10)) / 2) ** 2
    )

    return log_likelihood + log_prior


# Initialize the chain in log space
current = np.log([initial_a, initial_b])

current_posterior = log_posterior(current)

# Store MCMC samples
chain = np.zeros((n_iterations, 2))

accepted = 0

# Metropolis-Hastings sampling
for i in range(n_iterations):

    # Generate proposed values
    proposal = current + rng.normal(
        loc=0,
        scale=[0.08, 0.08],
        size=2
    )

    proposal_posterior = log_posterior(proposal)

    # Calculate acceptance probability
    log_acceptance = proposal_posterior - current_posterior

    # Accept or reject proposal
    if np.log(rng.random()) < min(0, log_acceptance):
        current = proposal
        current_posterior = proposal_posterior
        accepted += 1

    # Store current parameter values
    chain[i] = np.exp(current)


# Convert samples into a DataFrame
mcmc_df = pd.DataFrame(
    chain,
    columns=["a", "b"]
)

print("Total MCMC iterations:", n_iterations)
print("Accepted proposals:", accepted)
print("Acceptance rate:", round(accepted / n_iterations * 100, 2), "%")

display(mcmc_df.head())
```

```python
# Select three portions of the MCMC chain
sample_groups = {
    "First 50 Samples": mcmc_df.iloc[:50],
    "All Samples": mcmc_df,
    "Last 50 Samples": mcmc_df.iloc[-50:]
}

# Create trace plots
fig, axes = plt.subplots(2, 3, figsize=(16, 8))

for column_index, (group_name, samples) in enumerate(sample_groups.items()):

    # Plot shape parameter a
    axes[0, column_index].plot(
        samples["a"].to_numpy(),
        marker=".",
        linewidth=1
    )

    axes[0, column_index].set_title(
        f"{group_name} - Parameter a"
    )
    axes[0, column_index].set_xlabel("Iteration")
    axes[0, column_index].set_ylabel("Value of a")
    axes[0, column_index].grid(alpha=0.3)

    # Plot scale parameter b
    axes[1, column_index].plot(
        samples["b"].to_numpy(),
        marker=".",
        linewidth=1
    )

    axes[1, column_index].set_title(
        f"{group_name} - Parameter b"
    )
    axes[1, column_index].set_xlabel("Iteration")
    axes[1, column_index].set_ylabel("Value of b")
    axes[1, column_index].grid(alpha=0.3)

plt.suptitle("MCMC Trace Plots for Gamma Parameters")
plt.tight_layout()
plt.show()
```

# Conclusion

## The Metropolis-Hastings MCMC algorithm was implemented to estimate the Gamma distribution's shape parameter (a) and scale parameter (b). The chain was initialized with a = 4 and b = 10, and 5,000 iterations were generated. Trace plots were created for the first 50 samples, all samples, and the last 50 samples. These plots help examine the movement and behavior of the parameter estimates throughout the sampling process.

# Step 5: Visualizing MCMC Traces and Histograms After Burn-in

## **Objective:** To visualize the sampled values of parameters a and b and examine their posterior distributions after removing burn-in samples.

## In MCMC, burn-in refers to the initial iterations of a chain that may not yet represent the target posterior distribution.

## In this step:

### - We remove the first 1,000 iterations as burn-in.

### - The remaining samples are used for posterior analysis.

### - Trace plots show how parameters a and b change across iterations.

### - Histograms show the frequency distribution of the sampled parameter values.

## The trace plots help assess whether the chain has stabilized, while histograms provide information about the uncertainty and distribution of the estimated parameters.

## **Note:** Removing 1,000 iterations is a chosen burn-in setting for this practical. It should ideally be supported by convergence diagnostics rather than assumed to guarantee convergence.

```python
# Number of initial samples to remove
burn_in = 1000

# Remove burn-in samples
posterior_samples = mcmc_df.iloc[burn_in:].copy()

print("Total MCMC samples:", len(mcmc_df))
print("Burn-in samples removed:", burn_in)
print("Remaining posterior samples:", len(posterior_samples))

display(posterior_samples.head())
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Trace plot for parameter a
axes[0].plot(posterior_samples["a"])
axes[0].set_title("Trace Plot of Parameter a After Burn-in")
axes[0].set_xlabel("Iteration")
axes[0].set_ylabel("Value of a")
axes[0].grid(alpha=0.3)

# Trace plot for parameter b
axes[1].plot(posterior_samples["b"])
axes[1].set_title("Trace Plot of Parameter b After Burn-in")
axes[1].set_xlabel("Iteration")
axes[1].set_ylabel("Value of b")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Histogram of parameter a
axes[0].hist(
    posterior_samples["a"],
    bins=40,
    edgecolor="black",
    alpha=0.75
)

axes[0].set_title("Posterior Distribution of Parameter a")
axes[0].set_xlabel("Value of a")
axes[0].set_ylabel("Frequency")
axes[0].grid(alpha=0.3)

# Histogram of parameter b
axes[1].hist(
    posterior_samples["b"],
    bins=40,
    edgecolor="black",
    alpha=0.75
)

axes[1].set_title("Posterior Distribution of Parameter b")
axes[1].set_xlabel("Value of b")
axes[1].set_ylabel("Frequency")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

# Step 6: Prediction Using Estimated Parameters a and b

## **Objective:** To estimate the Gamma distribution parameters using the posterior MCMC samples and generate predicted sunspot observations.

## After removing burn-in samples, the remaining MCMC samples represent our estimated posterior distributions.

## We calculate:

### - **Posterior mean:** Average estimated value of the parameter.

### - **Posterior median:** Middle value of the posterior samples.

### - **Posterior standard deviation:** Measures uncertainty in the parameter estimates.

## The posterior mean values of a and b are used as estimates for the Gamma distribution.

## Finally, we generate simulated sunspot observations using the estimated parameters and compare their distribution with the actual sunspot observations.

## This helps us evaluate how well the fitted Gamma model represents the observed data.

```python
# Calculate posterior estimates
estimated_a = posterior_samples["a"].mean()
estimated_b = posterior_samples["b"].mean()

# Calculate posterior medians
median_a = posterior_samples["a"].median()
median_b = posterior_samples["b"].median()

# Calculate uncertainty
std_a = posterior_samples["a"].std()
std_b = posterior_samples["b"].std()

print("Estimated Gamma Parameters")
print("--------------------------")

print("Shape parameter (a):", round(estimated_a, 4))
print("Scale parameter (b):", round(estimated_b, 4))

print("\nPosterior Median")
print("Median of a:", round(median_a, 4))
print("Median of b:", round(median_b, 4))

print("\nPosterior Standard Deviation")
print("Standard deviation of a:", round(std_a, 4))
print("Standard deviation of b:", round(std_b, 4))
```

```python
from scipy.stats import gamma

# Generate predicted sunspot observations
predicted_sunspots = gamma.rvs(
    a=estimated_a,
    loc=0,
    scale=estimated_b,
    size=len(data),
    random_state=42
)

# Compare actual and predicted distributions
plt.figure(figsize=(10, 6))

plt.hist(
    data,
    bins=30,
    density=True,
    alpha=0.6,
    label="Actual Sunspot Counts"
)

plt.hist(
    predicted_sunspots,
    bins=30,
    density=True,
    alpha=0.6,
    label="Predicted Sunspot Counts"
)

plt.title("Actual vs Predicted Sunspot Distribution")
plt.xlabel("Sunspot Number")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
comparison = pd.DataFrame({
    "Measure": ["Mean", "Median", "Standard Deviation"],
    "Actual Data": [
        np.mean(data),
        np.median(data),
        np.std(data)
    ],
    "Predicted Data": [
        np.mean(predicted_sunspots),
        np.median(predicted_sunspots),
        np.std(predicted_sunspots)
    ]
})

display(comparison.round(3))
```
